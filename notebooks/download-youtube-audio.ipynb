```python
import subprocess
import time

def download_audio_with_retries(youtube_url, output_path, max_retries=5):
    retry_count = 0
    while retry_count < max_retries:
        try:
            command = [
                'yt-dlp', '-x', '--audio-format', 'mp3', '-o', output_path, youtube_url
            ]
            subprocess.run(command, check=True)
            return output_path
        except subprocess.CalledProcessError as e:
            print(f"Attempt {retry_count + 1} failed: {str(e)}")
            time.sleep(5)  # wait 5 seconds before retrying
            retry_count += 1
    print("Failed to download after several retries.")
    return None

# URL for the YouTube video
youtube_url = 'https://www.youtube.com/watch?v=MXSTNZ6yKbc'

# Download audio file
output_path = '104-MXSTNZ6yKbc.mp3'
audio_path = download_audio_with_retries(youtube_url, output_path)
if audio_path:
    print(f'Audio downloaded successfully: {audio_path}')
else:
    print('Audio download failed.')
```
